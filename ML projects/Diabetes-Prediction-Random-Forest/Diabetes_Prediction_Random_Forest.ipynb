```python
#  Copy dataset
df = std.copy()

# Handle invalid zero values
columns = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]

df[columns] = df[columns].replace(0, np.nan)

print("Missing Values:")
print(df.isnull().sum())
```
